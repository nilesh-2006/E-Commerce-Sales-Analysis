# Data Cleaning

## Objective

1. The objective of this notebook is to clean the raw dataset by handling missing values, duplicate records, incorrect data types, inconsistent column names, and invalid values.

2. The cleaned dataset will be used for feature engineering and business analysis.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/superstore.csv",encoding="latin1")
```

```python
df.head()
```

```python
# Check Dataset Shape
print("Dataset Shape :", df.shape)
```

```python
# Check Missing Values
df.isnull().sum()
```

```python
# Check Duplicate Records
df.duplicated().sum()
```

```python
# Remove Duplicate Records
df = df.drop_duplicates()

print("Dataset Shape After Removing Duplicates :", df.shape)
```

```python
# Check Data Types
df.dtypes
```

```python
# Convert Date Columns
df["Order Date"] = pd.to_datetime(df["Order Date"])

df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df[["Order Date", "Ship Date"]].dtypes
```

```python
# Remove Extra Spaces from Column Names
df.columns = df.columns.str.strip()
df.head()
```

```python
# Rename Columns (Snake Case)
df.columns = (
    df.columns
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("-", "_")
)
```

```python
df.columns
```

```python
# Check Negative Sales
df[df["sales"] < 0]
```

```python
# Check Negative Profit
df[df["profit"] < 0].head()
```

```python
# Check Invalid Quantity
df[df["quantity"] <= 0]
```

```python
# Check Discount Range
df["discount"].describe()
```

```python
# Check Postal Code Missing Values
df["postal_code"].isnull().sum()
```

```python
# Check Dataset Information
df.info()
```

```python
# Final Dataset Shape
print("Final Dataset Shape :", df.shape)
```
