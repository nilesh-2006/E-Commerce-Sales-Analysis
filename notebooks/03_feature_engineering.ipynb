# Feature Engineering

## Objective

1. The objective of this notebook is to create new features from the cleaned dataset.

2. These features will help perform better sales, profit, customer, product, and regional analysis.

```python
import pandas as pd
import numpy as np
```

# Load Clean Dataset

```python
df = pd.read_csv(
    "../data/processed/cleaned_superstore.csv",
    parse_dates=["order_date","ship_date"], encoding="latin1"
)
df.head()
```

- "parse_dates" isliye use kiya kyunki CSV me date dobara text ban jati hai. Isse Pandas load karte hi unhe datetime me convert kar deta hai

```python
# Convert Date Columns
df["order_date"] = pd.to_datetime(df["order_date"])
df["ship_date"] = pd.to_datetime(df["ship_date"])
df.head()
```

# Date Features

```python
# Year
df["year"] = df["order_date"].dt.year
df.head()
```

```python
# Month Number
df["month"] = df["order_date"].dt.month
df.head()
```

```python
# Month Name
df["month_name"] = df["order_date"].dt.month_name()
df.head()
```

```python
# Quarter 
df["quarter"] = df["order_date"].dt.quarter
df.head()
```

```python
# Weekday Name
df["day_name"] = df["order_date"].dt.day_name()
df.head()
```

```python
# Week Number
df["week"] = df["order_date"].dt.isocalendar().week
df.head()
```

```python
df["day"] = df["order_date"].dt.day
df[["order_date", "day"]].head()
```

# Shipping Features

```python
# Shipping Days
df["shipping_days"] = (
    df["ship_date"] - df["order_date"]
).dt.days
df.head()
```

# Business Features

```python
# Sales Per Unit
df["sales_per_unit"] = df["sales"] / df["quantity"]
df.head()
```

```python
# Profit Per Unit
df["profit_per_unit"] = df["profit"] / df["quantity"]
df.head()
```

```python
# Discount Amount
df["discount_amount"] = df["sales"] * df["discount"]
df.head()
```

```python
# create profit status
df["profit_status"] = df["profit"].apply(
    lambda x: "Loss" if x < 0 else "Profit"
)
df
```

```python
# Profit Margin (%)
df["profit_margin"] = (
    df["profit"] / df["sales"]
) * 100

df.head()
```

```python
# Order Value
df["order_value"] = df["sales"]
df.head()
```

```python
df.to_csv(
    "../data/processed/featured_superstore.csv",
    index=False
)
df.head()
```
